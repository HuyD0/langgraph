## 4. Big-O: what `O(n)` and `O(n²)` mean

**Big-O describes how the work grows when the input grows.** `n` is the size
of the input, such as the length of the list.

| Big-O | In words | Typical code |
|---|---|---|
| O(1) | Same work however big the input | `key in my_dict` |
| O(log n) | Halve the problem each step | binary search |
| O(n) | One step per item | one `for` loop |
| O(n log n) | A bit more than one pass | `sorted(nums)` |
| O(n²) | Every item against every item | a loop inside a loop |

Here's a function that checks whether a list contains a duplicate by comparing
every pair. For each item it loops over the rest of the list, so with `n` items
it does roughly `n × n` comparisons: **O(n²)**.

```python
def has_duplicate_slow(nums):
    """Compare every pair. Correct, but O(n²)."""
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] == nums[j]:
                return True
    return False


has_duplicate_slow([1, 2, 3, 2])
```

## Try it: watch Big-O happen

Run this. Each time `n` doubles, the O(n) version takes about twice as long,
and the O(n²) version takes about **four** times as long. That difference is
the whole point of Big-O: at small sizes both are fine, and at large sizes only
one of them finishes.

```python
for n in [1_000, 2_000, 4_000]:
    nums = list(range(n))  # no duplicates: the worst case, every pair gets checked

    start = time.perf_counter()
    has_duplicate_slow(nums)
    slow = time.perf_counter() - start

    start = time.perf_counter()
    has_duplicate(nums)
    fast = time.perf_counter() - start

    print(f"n={n:>5}   O(n²): {slow:.3f}s   O(n): {fast:.5f}s")
```
